```python
import pandas as pd
```

```python
df = pd.read_csv("cleaned_online_sales_data.csv")
```

```python
# Convert 'Order Date' back to datetime for time-based analysis
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

## Descriptive Statistics

```python
print(df[["Amount", "Profit", "Quantity"]].describe())
```

## Key Insights and Relationships

### Top sellers

```python
# 1. Top Categories by Amount and Profit
category_summary = df.groupby("Category")[["Amount", "Profit"]].sum().sort_values(by="Amount", ascending=False)
print("\nTop Categories by Amount and Profit:\n", category_summary)

# 2. Top Sub-Categories by Amount and Profit
subcategory_summary = df.groupby("Sub-Category")[["Amount", "Profit"]].sum().sort_values(by="Amount", ascending=False)
print("\nTop Sub-Categories by Amount and Profit:\n", subcategory_summary.head(10))
```

### Geographic Analysis

```python
# 3. Top States and Cities by profit
state_city_summary = df.groupby(["State", "City"])[["Amount", "Profit"]].sum().sort_values(by="Profit", ascending=False)
print("\nTop States and Cities by Amount and Profit:\n", state_city_summary.head(10))
```

### Time Analysis

```python
# 4. Sales and Profit Trends Over Time (by Month)
df["OrderMonth"] = df["Order Date"].dt.to_period("M")
monthly_trends = df.groupby("OrderMonth")[["Amount", "Profit"]].sum()
print("\nMonthly Sales and Profit Trends:\n", monthly_trends)

# 5. most profit months
most_profit_months = monthly_trends.sort_values(by="Profit", ascending=False)
print("\nMost Profitable Months:\n", most_profit_months.head(10))
```

### costumers & payment methods

```python
# 6. Most parchases costumers
top_customers = df.groupby("CustomerName")[["Amount", "Profit"]].sum().sort_values(by="Amount", ascending=False)
print("\nTop Customers by Amount and Profit:\n", top_customers.head(10))

# 7. Most payment modes used
payment_mode_count = df["PaymentMode"].value_counts()
print("\nMost Used Payment Modes:\n", payment_mode_count)

# 8. Profitability by Payment Mode
payment_mode_profit = df.groupby("PaymentMode")["Profit"].sum().sort_values(ascending=False)
print("\nProfitability by Payment Mode:\n", payment_mode_profit)
```

# Correlation & Hypothesis

```python
# 9. Correlation Matrix for numerical variables
print("\n--- Correlation Matrix ---")
print(df[["Amount", "Profit", "Quantity"]].corr())
```

```python
# 10. Hypothesis: Do certain categories have higher profit margins?
df["Profit_Margin"] = (df["Profit"] / df["Amount"]) * 100
category_profit_margin = df.groupby("Category")["Profit_Margin"].mean().sort_values(ascending=False)
print("\nAverage Profit Margin by Category:\n", category_profit_margin)
```

```python
# 11. Identify products with negative profit
negative_profit_items = df[df["Profit"] < 0].groupby(["Category", "Sub-Category"])[["Amount", "Profit", "Quantity"]].sum().sort_values(by="Profit", ascending=True)
print("\nItems with Negative Profit:\n", negative_profit_items.head(10))
```


## Business Insights and Recommendations

### Main Conclusions of Analysis:
1.  **Category Performance:** 'Electronics' and 'Clothing' are the top-performing categories in terms of both sales amount and profit. 'Furniture' lags behind, and notably, has a negative average profit margin.
2.  **Sub-Category Performance:** 'Printers', 'Saree', and 'Bookcases' are leading sub-categories by sales amount. However, several sub-categories, particularly within 'Electronics' (e.g., Phones, Electronic Games) and 'Clothing' (e.g., Saree), show significant negative profits.
3.   ability:** 'Credit Card' and 'COD' (Cash on Delivery) are the most profitable payment modes, indicating strong customer preference or higher transaction values through these methods.
4.  **Monthly Trends:** Sales and profit show fluctuations throughout the year, with a notable dip in profit in May and December, and a strong recovery in November. This suggests seasonality or specific events impacting sales.
5.  **Geographic Performance:** Madhya Pradesh (specifically Indore) and Maharashtra (Mumbai, Pune) are key regions contributing significantly to sales and profit.
6.  **Profit Margin Discrepancies:** While 'Clothing' and 'Electronics' have positive average profit margins, 'Furniture' has a negative average profit margin, indicating a systemic issue within this category.
7.  **Negative Profit Items:** A significant number of items across various categories (especially 'Electronics' and 'Clothing' sub-categories like 'Phones', 'Saree', 'Electronic Games') are generating negative profit, which is a critical area for intervention.

### Business Recommendations:
1.  **Optimize Product Portfolio:** Conduct a thorough review of products within 'Furniture' category and sub-categories identified with negative profits (e.g., 'Phones', 'Saree', 'Electronic Games', 'Bookcases'). Consider discontinuing consistently unprofitable items, renegotiating supplier costs, or adjusting pricing strategies.
2.  **Targeted Marketing:** Focus marketing efforts on high-performing categories ('Electronics', 'Clothing') and sub-categories ('Printers', 'Saree', 'Bookcases') to maximize revenue. Leverage insights from top-performing states and cities (Madhya Pradesh, Maharashtra) for geographically targeted campaigns.
3.  **Payment Mode Promotion:** Encourage the use of 'Credit Card' and 'COD' payment modes through promotions or incentives, as they are associated with higher profitability.
4.  **Seasonal Sales Strategies:** Analyze the reasons for profit dips in May and December. Implement targeted promotions or inventory adjustments during these months to mitigate losses and capitalize on potential opportunities.
5.  **Supplier and Cost Review:** Investigate the cost structure of products with negative profit margins. This could involve negotiating better deals with suppliers, optimizing logistics, or reducing operational overheads.
6.  **Customer Segmentation:** Further analyze customer behavior in profitable regions and for profitable products to identify key customer segments. Tailor marketing and product offerings to these segments.

### Limitations of Analysis and Next Steps:
1.  **Limited Data Granularity:** The current dataset lacks detailed customer demographics, marketing spend, and competitor data, which could provide deeper insights into sales drivers and market dynamics.
2.  **Short Timeframe:** The data appears to be for a single year (2018). A longer time series would allow for more robust trend analysis and seasonality identification.
3.  **External Factors:** The analysis does not account for external factors such as economic conditions, major events, or competitor activities that could influence sales and profit.
