```python
import pandas as pd
```

```python
#Load data
try:
    details_df = pd.read_csv('Details.csv')
    orders_df = pd.read_csv('Orders.csv')
except FileNotFoundError as e:
    print(f"Error: {e}. Please ensure the CSV files are in the correct directory.")
    raise
```

```python
#Data cleaning and preparation for Details.csv
print("Details.csv Info:")
print(details_df.info(),"\n-----------------------------") 
print("Missing values:\n",details_df.isnull().sum(),"\n-----------------------------")
print("Duplicate rows:\n",details_df.duplicated().sum())
```

```python
#Data cleaning and preparation for Orders.csv
print("Orders.csv Info:")
print(orders_df.info(),"\n-----------------------------") 
print("Missing values:\n",orders_df.isnull().sum(),"\n-----------------------------")
print("Duplicate rows:\n", orders_df.duplicated().sum())
```

```python
# Convert 'Order Date' to datetime objects
orders_df['Order Date'] = pd.to_datetime(orders_df['Order Date'], format='%d-%m-%Y', errors='coerce')
```

```python
merged_df = pd.merge(details_df, orders_df, on='Order ID', how='inner')
print("\n--- Merged DataFrame Info ---")
merged_df.info()
print("\nMissing values in Merged DataFrame:\n", merged_df.isnull().sum())
print("\nDuplicate rows in Merged DataFrame:", merged_df.duplicated().sum())

output_path = 'cleaned_online_sales_data.csv'
merged_df.to_csv(output_path, index=False)
print(f"\nCleaned and merged data saved to {output_path}")

print("\nFirst 5 rows of the cleaned and merged data:\n", merged_df.head())
```
